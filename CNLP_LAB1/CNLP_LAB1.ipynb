# Experiment-1: Data Analysis
### Dataset: `employee_information_100`

This notebook loads the employee dataset and performs the analysis tasks
requested in Experiment 1.1 to 1.10, including grouping, filtering, and
visualizations (bar chart, pie chart, histogram, scatter plot).

## Setup: Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

# Nicer default plot size
plt.rcParams['figure.figsize'] = (8, 5)
```

## Load Dataset

Update `DATA_PATH` below if your file is located elsewhere
(e.g. `/content/employee_information_100.csv` on Google Colab).

```python
DATA_PATH = "employee_information_100.csv"

df = pd.read_csv(DATA_PATH)

print("Shape of dataset:", df.shape)
df.head()
```

```python
# Quick look at data types and missing values
df.info()
```

## Experiment 1.1: Average Salary of Each Department (Bar Chart)

```python
avg_salary_dept = df.groupby('Department')['Salary'].mean().sort_values(ascending=False)
print(avg_salary_dept)

avg_salary_dept.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Average Salary by Department')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Experiment 1.2: Number of Employees in Each Department

```python
employee_count_dept = df['Department'].value_counts()
print(employee_count_dept)
```

## Experiment 1.3: Percentage of Male and Female Employees (Pie Chart)

```python
gender_counts = df['Gender'].value_counts()
print(gender_counts)

plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct='%1.1f%%',
    colors=['skyblue', 'lightpink'],
    startangle=90
)
plt.title('Male vs Female Employees')
plt.axis('equal')
plt.show()
```

## Experiment 1.4: Salary Distribution (Histogram)

```python
plt.hist(df['Salary'], bins=10, color='seagreen', edgecolor='black')
plt.title('Salary Distribution')
plt.xlabel('Salary')
plt.ylabel('Number of Employees')
plt.show()
```

## Experiment 1.5: Experience vs Salary (Scatter Plot)

```python
plt.scatter(df['Experience_Years'], df['Salary'], color='purple', alpha=0.7)
plt.title('Experience vs Salary')
plt.xlabel('Experience (Years)')
plt.ylabel('Salary')
plt.show()
```

## Experiment 1.6: Top 10 Employees with the Highest Salary

```python
top_10_salary = df.sort_values(by='Salary', ascending=False).head(10)
top_10_salary[['Employee_ID', 'Name', 'Department', 'Salary']]
```

## Experiment 1.7: Highest Salary in Every Department

```python
highest_salary_dept = df.groupby('Department')['Salary'].max().sort_values(ascending=False)
print(highest_salary_dept)
```

## Experiment 1.8: Employees Earning Above the Overall Average Salary

```python
overall_avg_salary = df['Salary'].mean()
print(f"Overall Average Salary: {overall_avg_salary:.2f}")

above_avg_employees = df[df['Salary'] > overall_avg_salary]
above_avg_employees[['Employee_ID', 'Name', 'Department', 'Salary']]
```

## Experiment 1.9: Average Years of Experience in Each Department

```python
avg_experience_dept = df.groupby('Department')['Experience_Years'].mean().sort_values(ascending=False)
print(avg_experience_dept)
```

## Experiment 1.10: Distribution of Employee Ages (Histogram)

```python
plt.hist(df['Age'], bins=10, color='teal', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.show()
```
