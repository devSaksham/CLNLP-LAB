## Experiment 2.1: Basic Text Preprocessing

```python
# Load the raw text
with open("2.1_text_data.txt", "r") as f:
    text = f.read()

print("----- Original Text -----")
print(text)
```

### Step 1: Convert to Lowercase (+ Uppercase Count)

```python
uppercase_count = sum(1 for ch in text if ch.isupper())

text_lower = text.lower()

print(f"Number of uppercase letters in original text: {uppercase_count}")
print("\n----- Text after Lowercasing -----")
print(text_lower)
```

### Step 2: Remove Punctuation (+ Types of Punctuation Used)

```python
import string

punctuation_found = sorted(set(ch for ch in text_lower if ch in string.punctuation))

text_no_punct = "".join(ch for ch in text_lower if ch not in string.punctuation)

print(f"Different types of punctuation found ({len(punctuation_found)}): {punctuation_found}")
print("\n----- Text after Removing Punctuation -----")
print(text_no_punct)
```

### Step 3: Remove Numbers (+ Count of Numbers)

```python
import re

numbers_found = re.findall(r"\d+", text_no_punct)

text_no_numbers = re.sub(r"\d+", "", text_no_punct)

print(f"Numbers found ({len(numbers_found)}): {numbers_found}")
print("\n----- Text after Removing Numbers -----")
print(text_no_numbers)
```

### Step 4: Remove Extra Whitespace (+ Amount of Extra Whitespace)

```python
extra_ws_matches = re.findall(r"\s{2,}", text_no_numbers)
extra_ws_chars = sum(len(m) - 1 for m in extra_ws_matches)

cleaned_text = re.sub(r"\s+", " ", text_no_numbers).strip()

print(f"Extra whitespace occurrences (2+ consecutive spaces/newlines): {len(extra_ws_matches)}")
print(f"Total extra whitespace characters removed: {extra_ws_chars}")
print("\n----- Text after Removing Extra Whitespace -----")
print(cleaned_text)
```

### Step 5: Display the Cleaned Text

```python
print("===== Final Cleaned Text =====")
print(cleaned_text)
```

## Experiment 2.2: Word & Sentence Tokenization

Three separate implementations: **NLTK**, **spaCy**, and **without any library**.

```python
with open("2.2_tokenization_data.txt", "r") as f:
    tokenization_text = f.read().strip()

print(tokenization_text)
```

### 2.2.1 Using NLTK

```python


import nltk
nltk.download("punkt")

from nltk.tokenize import sent_tokenize, word_tokenize

nltk_sentences = sent_tokenize(tokenization_text)
nltk_words = word_tokenize(tokenization_text)

print(f"NLTK Sentence Count: {len(nltk_sentences)}")
print("Sentences:")
for i, s in enumerate(nltk_sentences, 1):
    print(f"{i}. {s}")

print(f"\nNLTK Word Count: {len(nltk_words)}")
print("Words:")
print(nltk_words)
```

### 2.2.2 Using spaCy

```python

import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp(tokenization_text)

spacy_sentences = [sent.text.strip() for sent in doc.sents]
spacy_words = [token.text for token in doc if not token.is_space]

print(f"spaCy Sentence Count: {len(spacy_sentences)}")
print("Sentences:")
for i, s in enumerate(spacy_sentences, 1):
    print(f"{i}. {s}")

print(f"\nspaCy Word Count: {len(spacy_words)}")
print("Words:")
print(spacy_words)
```

### 2.2.3 Without NLTK / spaCy (Manual, Regex-based)

```python
import re

# Sentence tokenization: split after ., !, or ? followed by whitespace
manual_sentences = re.split(r"(?<=[.!?])\s+", tokenization_text.strip())
manual_sentences = [s.strip() for s in manual_sentences if s.strip()]

# Word tokenization: extract word-like tokens (letters/digits, keeps contractions/hyphens together)
manual_words = re.findall(r"\b\w+(?:[-']\w+)*\b", tokenization_text)

print(f"Manual Sentence Count: {len(manual_sentences)}")
print("Sentences:")
for i, s in enumerate(manual_sentences, 1):
    print(f"{i}. {s}")

print(f"\nManual Word Count: {len(manual_words)}")
print("Words:")
print(manual_words)
```

## Experiment 2.3: Stop Word Removal

```python
with open("2.3_clean_data.txt", "r") as f:
    stopword_text = f.read().strip()

print(stopword_text)
```

### Tokenize and Remove Stop Words

```python
import nltk
nltk.download("stopwords")
nltk.download("punkt")
nltk.download("punkt_tab")

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

original_tokens = word_tokenize(stopword_text)

stop_words = set(stopwords.words("english"))

filtered_tokens = [w for w in original_tokens if w.lower() not in stop_words]
stopwords_found = [w for w in original_tokens if w.lower() in stop_words]

print(f"Original Tokens ({len(original_tokens)}):")
print(original_tokens)

print(f"\nFiltered Tokens - stopwords removed ({len(filtered_tokens)}):")
print(filtered_tokens)

print(f"\nStop Words Found in Text ({len(stopwords_found)}):")
print(stopwords_found)
```
