## Experiment 3.1 — Stemming using Porter Stemmer

Stemming reduces a word to its root/stem form by chopping off suffixes, using fixed rules. The output stem is **not always a valid dictionary word** (e.g. "studies" → "studi").

```python
# Words to stem
words_stemming = [
    "playing", "played", "plays", "studies",
    "studying", "connected", "connection", "computers"
]

# Initialize Porter Stemmer
porter = PorterStemmer()

# Apply stemming
stemming_results = [(word, porter.stem(word)) for word in words_stemming]

# Display results
print(f"{'Original Word':<15}{'Stemmed Word':<15}")
print("-" * 30)
for original, stemmed in stemming_results:
    print(f"{original:<15}{stemmed:<15}")
```

## Experiment 3.2 — Lemmatization using WordNet Lemmatizer

Lemmatization reduces a word to its dictionary/base form (the **lemma**) using vocabulary and morphological analysis, so the output is always a valid word.

Lemmatization quality depends on knowing the word's **Part of Speech (POS)** — e.g. "leaves" lemmatizes differently as a noun ("leaf") than as a verb ("leave"). We first POS-tag each word, map the POS tag to the WordNet format, then lemmatize accordingly.

```python
def get_wordnet_pos(treebank_tag):
    """Map POS tag (from nltk.pos_tag) to the format WordNetLemmatizer expects."""
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN  # default fallback
```

```python
# Words to lemmatize
words_lemmatization = [
    "cats", "dogs", "running", "runs", "ran",
    "studies", "studying", "better", "children",
    "mice", "went", "ate", "leaves", "caring"
]

# Initialize WordNet Lemmatizer
lemmatizer = WordNetLemmatizer()

# POS-tag the words (needed for accurate lemmatization)
tagged_words = pos_tag(words_lemmatization)

# Apply lemmatization using the mapped POS tag
lemmatization_results = [
    (word, tag, lemmatizer.lemmatize(word, get_wordnet_pos(tag)))
    for word, tag in tagged_words
]

# Display results
print(f"{'Original Word':<15}{'POS Tag':<10}{'Lemma':<15}")
print("-" * 40)
for original, tag, lemma in lemmatization_results:
    print(f"{original:<15}{tag:<10}{lemma:<15}")
```

**Note:** `pos_tag` guesses the part of speech from context. Since each word here is tagged in isolation (not inside a sentence), it can occasionally mistag a word — e.g. "ate" may be tagged as an adjective instead of a verb, which affects its lemma. This is a real limitation of POS taggers and a good discussion point in the experiment report.

## Experiment 3.3 — Information Extraction using Regular Expressions

Extract the following entities from the input text using regex:
- Email addresses
- URLs
- Mobile numbers
- Hashtags
- Mentions (@username)

```python
# Input text
text = (
    "Welcome to the Natural Language Processing workshop! For more information, "
    "visit https://www.nlpworkshop.com or www.python.org. "
    "You can contact the coordinator at nlpworkshop@gmail.com or support@python.org. "
    "For registration, call +91-9876543210 or 9123456789. "
    "Follow us on social media @NLPWorkshop and @PythonLearner. "
    "Share your experience using #NLP, #Python, and #MachineLearning. "
    "You can also visit https://github.com/NLPWorkshop for the latest updates."
)

print(text)
```

```python
# Regex patterns
email_pattern   = r'[\w.-]+@[\w.-]+\.\w+'
url_pattern     = r'https?://[^\s]+|www\.[^\s,]+'
mobile_pattern  = r'(?:\+91[-\s]?)?\d{10}'
hashtag_pattern = r'#\w+'
# Negative lookbehind ensures '@gmail' inside an email is NOT matched as a mention
mention_pattern = r'(?<![\w.])@\w+'
```

```python
# Extract each entity type
emails   = re.findall(email_pattern, text)
urls     = [u.rstrip('.,)') for u in re.findall(url_pattern, text)]  # strip trailing punctuation
mobiles  = re.findall(mobile_pattern, text)
hashtags = re.findall(hashtag_pattern, text)
mentions = re.findall(mention_pattern, text)
```

```python
# Display extracted results
print("Email Addresses:", emails)
print("URLs:            ", urls)
print("Mobile Numbers:  ", mobiles)
print("Hashtags:        ", hashtags)
print("Mentions:        ", mentions)
```
